```python
import pandas as pd

# Charger le fichier CSV
df = pd.read_csv('/content/Data Warehouse Multiclass.csv')

# Afficher les premières lignes
display(df.head())
```

```python
# Afficher les informations sur les colonnes et types de données
df.info()
```

### Identification des variables $X$ (Caractéristiques) et $Y$ (Cible)

```python
# 1. Lister toutes les colonnes disponibles
print("Colonnes disponibles dans le dataset :\n", df.columns.tolist())

# 2. Analyser le nombre de valeurs uniques par colonne
# Une colonne avec peu de valeurs uniques est souvent un bon candidat pour être un Y de classification.
print("\nNombre de valeurs uniques par colonne :")
for col in df.columns:
    nunique = df[col].nunique()
    dtype = df[col].dtype
    print(f"- {col} ({dtype}) : {nunique} valeurs uniques")
```

Généralement, pour définir vos variables :
- **$Y$ (Cible)** : C'est la variable que vous voulez prédire (par exemple, une classe, un label, ou une valeur numérique).
- **$X$ (Caractéristiques)** : Ce sont toutes les autres colonnes utiles pour la prédiction (en excluant les identifiants uniques comme les IDs si présents).

```python
# Exemple générique de séparation en X et Y
# Remplacer 'NOM_DE_VOTRE_COLONNE_CIBLE' par la colonne réelle que vous choisissez comme cible (Y)
target_column = df.columns[-1]  # Par défaut, on prend la dernière colonne (label)

Y = df[target_column]
X = df.drop(columns=[target_column])

print("Exemple configuré :\n- Variable cible (Y) :", target_column, "\n- Variables prédictives (X) :", X.columns.tolist()[:10], "... (total:", len(X.columns), "colonnes)")
```

### Analyse et Explication des variables $X$ et $Y$

```python
# 1. Analyse descriptive de la variable cible Y
print("=== ANALYSE DE LA CIBLE (Y) ===")
print(Y.describe())
print("\nDistribution des classes de Y (Fréquences) :")
print(Y.value_counts(dropna=False))
print("\nDistribution en pourcentage :")
print(Y.value_counts(normalize=True, dropna=False) * 100)
```

```python
# 2. Analyse descriptive des variables explicatives X
print("=== ANALYSE DES CARACTÉRISTIQUES (X) ===")
# Résumé des types de données et valeurs manquantes pour X
print(X.info())

# Statistiques descriptives pour les variables numériques
numerical_cols = X.select_dtypes(include=['number']).columns
if len(numerical_cols) > 0:
    print("\nStatistiques des variables numériques :")
    display(X[numerical_cols].describe())

# Statistiques descriptives pour les variables catégorielles
categorical_cols = X.select_dtypes(exclude=['number']).columns
if len(categorical_cols) > 0:
    print("\nStatistiques des variables catégorielles :")
    display(X[categorical_cols].describe())
```

### Visualisation Graphique de $X$ et $Y$

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration du style des graphiques
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 5)
```

```python
# 1. Graphique de la distribution de la variable cible Y
plt.figure(figsize=(8, 5))
sns.countplot(x=Y, palette="viridis")
plt.title(f"Distribution de la variable cible (Y : {Y.name})")
plt.xlabel(Y.name)
plt.ylabel("Nombre d'occurrences")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2. Distribution des variables numériques de X (si elles existent)
numeric_cols = X.select_dtypes(include=['number']).columns
if len(numeric_cols) > 0:
    # On limite à max 4 variables pour l'affichage initial
    cols_to_plot = numeric_cols[:4]
    fig, axes = plt.subplots(nrows=1, ncols=len(cols_to_plot), figsize=(4 * len(cols_to_plot), 4))
    if len(cols_to_plot) == 1:
        axes = [axes]
    for i, col in enumerate(cols_to_plot):
        sns.histplot(X[col], kde=True, ax=axes[i], color="skyblue")
        axes[i].set_title(f"Distribution de {col}")
    plt.tight_layout()
    plt.show()
else:
    print("Aucune variable numérique dans X à afficher.")
```

### Analyse Biologique & Médicale des Biomarqueurs

Nous allons utiliser des graphiques croisés pour analyser la distribution et la corrélation de biomarqueurs physiologiques critiques selon l'état clinique de la cible (Normal vs Abnormal) :

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration de la figure pour afficher 3 aspects biologiques clés
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Glucose vs HbA1c (Régulation glycémique)
sns.scatterplot(data=df.sample(n=1000, random_state=42), x='HbA1c_level', y='glucose', hue='label', alpha=0.7, ax=axes[0], palette='coolwarm')
axes[0].set_title("Régulation Glycémique\n(HbA1c vs Glucose)")
axes[0].set_xlabel("Taux d'HbA1c (%)")
axes[0].set_ylabel("Glucose (mg/dL)")

# 2. Pression Artérielle (Systolique vs Diastolique)
sns.scatterplot(data=df.sample(n=1000, random_state=42), x='systolic_bp', y='diastolic_bp', hue='label', alpha=0.7, ax=axes[1], palette='coolwarm')
axes[1].set_title("Profil Tensionnel\n(Systolique vs Diastolique)")
axes[1].set_xlabel("Pression Systolique (mmHg)")
axes[1].set_ylabel("Pression Diastolique (mmHg)")

# 3. Profil Lipidique (HDL vs LDL)
sns.scatterplot(data=df.sample(n=1000, random_state=42), x='hdl', y='ldl', hue='label', alpha=0.7, ax=axes[2], palette='coolwarm')
axes[2].set_title("Bilan Lipidique\n(HDL 'Bon' vs LDL 'Mauvais' Cholestérol)")
axes[2].set_xlabel("HDL (mg/dL)")
axes[2].set_ylabel("LDL (mg/dL)")

plt.tight_layout()
plt.show()
```
